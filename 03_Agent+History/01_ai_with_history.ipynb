```python
## Message History

# wrap model and make it stateful.
# track inputs and outputs of model, store them in db
# in future chat, it will load the messages and pass them into chain as part of input
```

```python
%pip install langchain_community
```

```python
from langchain_community.chat_message_histories import ChatMessageHistory
from langchain_core.chat_history import BaseChatMessageHistory
from langchain_core.runnables.history import RunnableWithMessageHistory

#store all sessiosn
store = {}

def get_session_history(session_id: str)-> BaseChatMessageHistory:
    # create history wrt to the session id

    if session_id not in store:
        #whatever chat happens, we store it in the store's session id
        #here we load the previous chat history if it exists, and return it
        store[session_id]  = ChatMessageHistory()
        
    return store[session_id]

#get the entire chat history to the model. interact with model based on history
with_message_history = RunnableWithMessageHistory( model, get_session_history)
```

```python
#dummy
config = {"configurable": {"session_id": "chat1"}}
```

```python
response = with_message_history.invoke(
    [
        HumanMessage(content = "Hi, I am the user and I am a Senior Data Scientist")
    ],
    config = config
)
```

```python
response.content
```

```python
with_message_history.invoke(
    [
        HumanMessage(content = "Who am I?")
    ],
    config = config
)
```

```python
## change the session id to chat2, so that we can have a new chat history
config1= {"configurable": {"session_id": "chat2"}}

response = with_message_history.invoke(
    [
        HumanMessage(content = "Who am I?")
    ],
    config = config1
)

response.content

# as the session id is different, the model will not have any previous history and will not be able to answer the question "Who am I?"
```

```python
response = with_message_history.invoke(
    [
        HumanMessage(content = "My Name is John")
    ],
    config = config1
)

response.content
```
