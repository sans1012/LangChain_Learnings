```python
from langchain_core.documents import Document

#document shows unit of text and associated metadata.. it has page and metadata

documents = [
    Document(
        page_content= 'Dogs are great pets. They are loyal and friendly.',
        metadata={'source': 'mammals-pet-doc'},
    ),

    Document(
            page_content= 'Goldfish are popular pets for begineers, requiring relatively simple care',
            metadata={'source': 'fish-pet-doc'},
        ),

    Document(
            page_content= 'Cats are independent and low-maintenance pets, but they can be aloof at times.',
            metadata={'source': 'mammals-pet-doc'},
        ),

    Document(
            page_content= 'Parrots are intelligent brids capable of mimicking human speech.',
            metadata={'source': 'birds-pet-doc'},
        ),

    Document(
            page_content= 'Rabbits are social animals that need plenty of space to hop around',
            metadata={'source': 'mammals-pet-doc'},
        ),

]
```

```python
documents
```

```python
# vector store

# convert text to word vectors/embeddings

from langchain_chroma import Chroma
import os
from dotenv import load_dotenv
load_dotenv()
groq_api_key = os.environ.get('GROQ_API_KEY')
os.environ['HF_TOKEN'] = os.getenv('HF_TOKEN')
```

```python
from langchain_groq import ChatGroq

llm = ChatGroq(groq_api_key=groq_api_key,
                model="openai/gpt-oss-20b", temperature=0.5)
```

```python
!pip install langchain_huggingface --q
```

```python
from langchain_huggingface import HuggingFaceEmbeddings

embeddings = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")
```

```python
vectorstore = Chroma.from_documents(documents, embedding = embeddings)

vectorstore.similarity_search("What are the best pets for begineers?")[0]
```

```python
#async query

await vectorstore.asimilarity_search("How are cats")
```

```python
vectorstore.similarity_search_with_score("How are cats")
```

```python
## retrievers - batch

from typing import List
from langchain_core.documents import Document
from langchain_core.runnables import RunnableLambda

retriever = RunnableLambda(vectorstore.similarity_search).bind(k = 1)
retriever.batch(['dog', 'cat'])
```

```python
## retrievers - implement as retriever method

from typing import List
from langchain_core.documents import Document
retriever = vectorstore.as_retriever(
    search_type = 'similarity',
    search_kwargs = {'k': 1}

)

retriever.batch(['dog', 'cat'])
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.runnables import RunnablePassthrough

message = """Answer this question using provided context only
        {question}
        Context:
        {context}
"""

prompt = ChatPromptTemplate.from_messages([('human', message)])

rag_chain = {
    "context": retriever,
    "question": RunnablePassthrough()
} | prompt | llm

response = rag_chain.invoke('What are the best pets for begineers?')
```
