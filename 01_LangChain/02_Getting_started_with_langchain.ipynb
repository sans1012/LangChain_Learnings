```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ['OPENAI_API_KEY'] = os.getenv('OPENAI_API_KEY')

os.environ['LANGCHAIN_API_KEY'] = os.getenv('LANGCHAIN_API_KEY')

os.environ['LANGCHAIN_TRACING_V2'] = "true"

os.environ['LANGCHAIN_PROJECT'] = os.getenv('LANGCHAIN_PROJECT')
```

```python
from langchain_openai import ChatOpenAI
llm  = ChatOpenAI(model = 'gpt-4o')

print(llm)
```

```python
# input and get output of llm

result = llm.invoke(" What is the france history?")
result
```

```python
### ChatPrompt Template
# Tells the LLm model how to behave
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_messages(
    [
        ("system", "Your are an expert Data scientist. Provide the logical solution to query"),
        ("user", "{input}")

    ]
)
prompt
```

```python
chain = prompt | llm

response = chain.invoke({"input": 'tell about langsmith'})
print(response)
```
